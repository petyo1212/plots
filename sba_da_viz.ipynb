### Data Analysis and Visualization with Python


```python
# Import libraries 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read Data
df=pd.read_csv(r'data\woocommerce-product-export.csv')
```

```python
#EDA: Clean Data
# Print info()
print(df.info())
```

```python
# Summary of Statistics 
print(df.describe())
```

```python
# Print first 5 rows of the dataset
print(df.head())
```

```python
# Print last 5 rows of the dataset
print(df.tail())
```

```python
# Print total_profit and month_number columns as subsets
subset = df[['total_profit', 'month_number']]
print(subset)
```

```python
# Get the x and y values for plotting
x= df['month_number']
y= df['total_profit']
```

```python
# Generate bar plot. Set title, x,y labels, grid, and y limit. Show the plot.
plt.figure(figsize=(10,6))
plt.bar(x, y)
plt.title('Company profit per month')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.grid(linestyle='--', alpha=0.5)
plt.ylim(0, 500000)
plt.show()
```

```python
# Read total profit and month. Show a line plot. style marker, legend location, grid, and y limit. Show the plot.
plt.figure(figsize=(10,6))
plt.plot(x, y, marker='o', markerfacecolor='black',  linestyle='--', linewidth=2, color='red', label='Profit data of last year')
plt.title('Company profit per month')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.grid(linestyle='--', alpha=0.5)
plt.ylim(100000, 500000)
plt.xticks(np.arange(1, 13, 1))
plt.legend(loc='lower right')
plt.show()
```

```python
# Print all sales data and show multiline plot. Units sold per month.
Product_names=['Face Cream', 'Face Wash', 'Toothpaste', 'Bathingsoap', 'Shampoo', 'Moisturizer']
for i in range (1, 7):
    plt.plot(df['month_number'], df[df.columns[i]], marker='o', label=f'{Product_names[i-1]}')
plt.xlabel('Month Number')
plt.ylabel('Sales units in Number')
plt.title('Sales Data')
plt.legend()
plt.ylim(0, 18000)
plt.xticks(df['month_number'])
plt.yticks(np.array([1000,2000, 4000, 6000, 8000, 10000, 12000, 15000, 18000]))
plt.show()
# Note: The above code generates a multiline plot for the sales data of different products over the months. Each product's sales are represented by a different line on the graph, allowing for easy comparison of sales trends across products.
# Moisturizer and Face Wash data overlap. To fix this, we can use a different marker for each product line to distinguish them better. 
# That will produce a different graph from what we are requested on SBA
```

```python
# Plot bathingsoap sales data for each moth and produce scatter plot. Set title, x,y labels, grid, and y limit. Show the plot.
plt.figure(figsize=(10,6))
plt.scatter(df['month_number'], df['bathingsoap'], color='blue', label='bathing soap Sales data')
plt.title('bathingsoap Sales data')
plt.xlabel('Month Number')  
plt.ylabel('Number of units Sold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(df['month_number'])
plt.ylim(5800, 15000)
plt.legend(loc='upper left')
plt.show()
```

```python
# Task one. Given data set below cretea line chart x=date, y=temperature. Set title, x,y labels, grid, and y limit. Show the plot.
date=["25/12","26/12","27/12"]
temp=[8.5,10.5,6.8]
plt.figure(figsize=(10,6))
plt.plot(date, temp, marker='o', markerfacecolor='black',  linestyle='--', linewidth=2, color='red', label='Temperature data')
plt.title('Date-wise Temperature')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.ylim(5, 12)
plt.xticks(date)
plt.legend(loc='upper left')
plt.show()
```

```python
# Task two. Given data set below cretea line chart x=weight, y=height. Set title, x,y labels, grid, and y limit. Show the plot.
height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]
weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]
plt.figure(figsize=(8,6))
plt.plot(weight, height, marker='o', markersize=10, markerfacecolor='green',  linestyle='-.', linewidth=2, color='green', label='Height vs Weight data')
plt.title('Average weight with respect to average height')
plt.xlabel('Weight in kg')
plt.ylabel('Height in cm')
plt.grid(True, linestyle='--', alpha=0.5)
plt.ylim(120, 165)
plt.legend(loc='lower right')
plt.show()
```
